# Reproduction Testing 

This notebook verifies whether the project can be reproduced from a clean setup.

It checks the environment, required files, automated tests, evaluation outputs, and generated plots.

The goal is to confirm that another user can run the project and obtain the expected testing and visualization results.

## 1. Environment Check

In this section, we verify that the notebook is running in a valid Python environment and that the project dependencies can be imported successfully.

```python
import sys

print("Python Version:")
print(sys.version)
```

### Dependency Verification

The following libraries are required by the project and should be available in the current environment.

```python
packages = [
    "pandas",
    "numpy",
    "pm4py",
    "sklearn",
    "matplotlib",
    "pytest"
]

for package in packages:
    try:
        __import__(package)
        print(f"[OK] {package} imported successfully")
    except ImportError:
        print(f"[ERROR] {package} is missing")
```

## 2. Repository and File Verification

In this section, we verify that the important project files are available in the repository.

This includes evaluation output files, visualization scripts, generated plots, and all automated test files.

```python
from pathlib import Path

# Project root is one level above notebooks/
PROJECT_ROOT = Path("..")

required_files = [
    "requirements.txt",
    "model_metrics.csv",
    "model_scores.csv",
    "remaining_time/visualization.py",
    "outputs/plots/mae_vs_prefix_length.png",
    "outputs/plots/rmse_vs_prefix_length.png"
]

print("Checking required project files:\n")

for file in required_files:
    file_path = PROJECT_ROOT / file

    if file_path.exists():
        print(f"[OK] {file}")
    else:
        print(f"[MISSING] {file}")

# Detect all pytest files
test_files = sorted((PROJECT_ROOT / "tests").glob("test_*.py"))

print("\nDetected automated test files:\n")

if test_files:
    for file in test_files:
        print(f"[OK] {file.name}")
else:
    print("[MISSING] No pytest test files were found.")
```

## 3. Automated Test Execution

In this section, we run the automated pytest test suite to verify that the core project components work correctly.

```python
import subprocess

result = subprocess.run(
    ["python", "-m", "pytest", "../tests"],
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode == 0:
    print("\n[OK] All automated tests passed successfully.")
else:
    print("\n[ERROR] Some tests failed.")
    print(result.stderr)
```

## 4. Plot Verification

The project generates visualization outputs for model evaluation.

This step verifies that the expected plot files have been generated successfully.

```python
from pathlib import Path

plot_files = [
    "../outputs/plots/mae_vs_prefix_length.png",
    "../outputs/plots/rmse_vs_prefix_length.png"
]

print("Checking generated plots:\n")

for plot in plot_files:
    if Path(plot).exists():
        print(f"[OK] {plot}")
    else:
        print(f"[MISSING] {plot}")
```

## 5. Conclusion

The reproduction testing procedure was completed successfully.

The project environment was verified, all required dependencies were available, the repository files were detected correctly, the automated test suite passed successfully, and the expected visualization outputs were found.

Based on these results, the project can be reproduced successfully from a clean setup and is ready for deployment, further development, and evaluation.
